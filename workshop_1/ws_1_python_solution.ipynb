## Plot 3

```python
###### Stacked Bar Chart
# Create the input vectors
colors = ["green", "orange", "brown"]
months = ["Mar", "Apr", "May", "Jun", "Jul"]
regions = ["East", "West", "North"]

# Create the matrix of the values
Values = np.array([
    [2, 9, 3, 11, 9],
    [4, 8, 7, 3, 12],
    [5, 2, 8, 10, 11],
])

# Plot the stacked bar chart
plt.barh(months, Values[0], color=colors[0])
plt.barh(months, Values[1], left=Values[0], color=colors[1])
plt.barh(months, Values[2], left=Values[0] + Values[1], color=colors[2])

# Add labels and title
plt.xlabel("Revenue")
plt.ylabel("Month")
plt.title("Sales Revenue")

# Add legend
plt.legend(regions, loc="lower right", fontsize=13)

# Display the chart
plt.show()
```

**Plot 3:** Move the legend to topleft and make the stacked chart vertical

```python
colors = ["green", "orange", "brown"]
months = ["Mar", "Apr", "May", "Jun", "Jul"]
regions = ["East", "West", "North"]

Values = np.array([
    [2, 9, 3, 11, 9],
    [4, 8, 7, 3, 12],
    [5, 2, 8, 10, 11],
])

plt.bar(months, Values[0], color=colors[0])
plt.bar(months, Values[1], bottom=Values[0], color=colors[1])
plt.bar(months, Values[2], bottom=Values[0] + Values[1], color=colors[2])

plt.xlabel("Month")
plt.ylabel("Revenue")
plt.title("Sales Revenue")
plt.legend(regions, loc="upper left", fontsize=13)
plt.show()
```

## Plot 5

```python
##### Boxplot
# Generate some normally distributed univariate data
np.random.seed(1)
distr_data = np.random.normal(20.5, 5.5, 1000)

# Basic boxplot
plt.boxplot(distr_data)
plt.show()

# Boxplot with notches
plt.boxplot(distr_data, notch=True)
plt.show()
```

**Plot 5:** Change the seed from 1 to 42, how does it change the plot? Comment the answer

```python
# Seed 42 draws a different sample from the same normal distribution
# (mean 20.5, sd 5.5). The box, median, whiskers, and outliers shift.
# The overall shape stays similar.
np.random.seed(42)
distr_data = np.random.normal(20.5, 5.5, 1000)

plt.boxplot(distr_data)
plt.show()

plt.boxplot(distr_data, notch=True)
plt.show()
```

## Plot 6

```python
##### Histogram
np.random.seed(1)
distr_data = np.random.normal(20.5, 5.5, 1000)

# Basic histogram
plt.hist(distr_data)
plt.show()

# Histogram with a different number of bins
plt.hist(distr_data, bins=20)
plt.show()
```

**Plot 6:** Change the number of bins to 5 and the border to black, how did the plot change with the number of bins

```python
# Five bins are wider than the default histogram and the 20-bin histogram.
# Nearby values merge into fewer bars, so the shape looks coarser.
plt.hist(distr_data, bins=5, edgecolor="black")
plt.show()
```

## Plot 7

```python
##### Density Plot
np.random.seed(1)
distr_data = np.random.normal(20.5, 5.5, 1000)

# Generate density plot
density = gaussian_kde(distr_data)
x = np.linspace(min(distr_data), max(distr_data), 1000)
plt.plot(x, density(x), label="Density")
plt.fill_between(x, density(x), color="red", edgecolor="blue")
plt.title("Sale Amount Distribution")
plt.show()
```

**Plot 7:** Change the colour of the plot to purple

```python
density = gaussian_kde(distr_data)
x = np.linspace(min(distr_data), max(distr_data), 1000)
plt.plot(x, density(x), color="purple", label="Density")
plt.fill_between(x, density(x), color="purple", edgecolor="purple")
plt.title("Sale Amount Distribution")
plt.show()
```

## Plot 8

```python
##### Beeswarm Plot
np.random.seed(1)
distr_data = np.random.normal(20.5, 5.5, 1000)

# Beeswarm plot
sns.swarmplot(distr_data)
plt.show()

# Customize orientation, color, and symbol
sns.swarmplot(distr_data, color="red", marker="^", size=4, orient="h")
plt.show()
```

**Plot 8:** Change the orientation to horizontal and the plot points to a hollow triangle

```python
sns.swarmplot(
    distr_data,
    orient="h",
    marker="^",
    facecolor="none",
    edgecolor="black",
    linewidth=1,
    size=4,
)
plt.show()
```

## Plot 9

```python
##### Violin Plot
# Load the dataset from the GitHub URL
url = "https://raw.githubusercontent.com/vincentarelbundock/Rdatasets/master/csv/datasets/ToothGrowth.csv"
data = pd.read_csv(url)

# Convert 'dose' to a categorical variable
data["dose"] = data["dose"].astype("category")

# Prepare the data for violin plot
grouped_data = [group["len"].values for _, group in data.groupby("dose", observed=True)]

# Create the violin plot and beehive (individual points)
plt.violinplot(grouped_data, showmeans=False, showmedians=True)
for i, group in enumerate(grouped_data, start=1):
    plt.scatter([i] * len(group), group, color="black", s=5, alpha=0.6)

# Customize axis labels
plt.xticks([1, 2, 3], ["0.5", "1.0", "2.0"])
plt.xlabel("Dose")
plt.ylabel("Tooth Length")
plt.title("Violin Plot of Tooth Growth by Dose")
plt.show()
```

**Plot 9:** Add a boxplot with width 0.1 to the beehive

```python
plt.violinplot(grouped_data, showmeans=False, showmedians=True)
plt.boxplot(grouped_data, positions=[1, 2, 3], widths=0.1)

for i, group in enumerate(grouped_data, start=1):
    plt.scatter([i] * len(group), group, color="black", s=5, alpha=0.6)

plt.xticks([1, 2, 3], ["0.5", "1.0", "2.0"])
plt.xlabel("Dose")
plt.ylabel("Tooth Length")
plt.title("Violin Plot of Tooth Growth by Dose")
plt.show()
```

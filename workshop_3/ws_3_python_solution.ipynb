```python
# Import necessary libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Load sample datasets
mtcars = sns.load_dataset('mpg').dropna()
iris = sns.load_dataset('iris')
```

##### Basic Scatter Plot


```python
# Scatter Plot - Base Plot
x = [6,7,8,7,2,3,9,4,10,15,9,7]
y = [98,85,87,88,112,104,87,94,78,78,85,86]

# Basic Scatter Plot
plt.figure(figsize=(5,3))
plt.scatter(x, y)
plt.title("Basic Scatter Plot")
plt.show()
```

**Plot 1:** Repeat the command and try a few colours (for example `"coral"`).


```python
# Scatter Plot - Base Plot
x = [6,7,8,7,2,3,9,4,10,15,9,7]
y = [98,85,87,88,112,104,87,94,78,78,85,86]

# Basic Scatter Plot
plt.figure(figsize=(5,3))
plt.scatter(x, y, color="coral")
plt.title("Basic Scatter Plot")
plt.show()
```

```python
# Scatter Plot with Blue Points
x = [6,7,8,7,2,3,9,4,10,15,9,7]
y = [98,85,87,88,112,104,87,94,78,78,85,86]

plt.figure(figsize=(5,3))
plt.scatter(x, y, color="blue")
plt.show()
```

**Plot 2:** Try it again, but with a different point type and a fill colour gold.


```python
# Different point type with gold fill
x = [6,7,8,7,2,3,9,4,10,15,9,7]
y = [98,85,87,88,112,104,87,94,78,78,85,86]

plt.figure(figsize=(5,3))
plt.scatter(x, y, edgecolors="blue", facecolors="gold", marker="s")
plt.show()
```

##### Scatter Plot using seaborn


```python
# Scatter plot using seaborn
sns.scatterplot(data=mtcars, x="weight", y="mpg", s=100)
plt.show()
```

**Plot 3:** Change the points shape to diamond and make them twice as big.


```python
# Diamond markers, twice the default size (default s=36, so s=72)
sns.scatterplot(data=mtcars, x="weight", y="mpg", marker="D", s=200)
plt.show()
```

##### Scatter Plot with Colour and Fill


```python
sns.scatterplot(data=mtcars, x="weight", y="mpg", color="gold", edgecolor="blue", s=100)
plt.show()
```

**Plot 4:** Change the points shape to square with a red outline.


```python
sns.scatterplot(data=mtcars, x="weight", y="mpg", color="gold", edgecolor="red", marker="s", s=100)
plt.show()
```

##### Histogram


```python
# Histogram of horsepower (using 'horsepower' in seaborn mpg)
plt.figure(figsize=(6,4))
plt.hist(mtcars["horsepower"], bins=12, color="coral")
plt.xlabel("Horsepower")
plt.ylabel("Frequency")
plt.show()
```

**Plot 5:** Make the histogram 30% opacity using `alpha`.


```python
# Histogram of horsepower at 30% opacity
plt.figure(figsize=(6,4))
plt.hist(mtcars["horsepower"], bins=12, color="coral", alpha=0.3)
plt.xlabel("Horsepower")
plt.ylabel("Frequency")
plt.show()
```

##### Histogram with Transparency


```python
# Using alpha (transparency)
plt.figure(figsize=(6,4))
plt.hist(mtcars["horsepower"], bins=12, color="blue", alpha=0.3)
plt.show()
```

**Plot 6:** Add colour `"blue"` (edge colour) to the 30% opacity plot.


```python
# Using alpha (transparency) with blue edge colour
plt.figure(figsize=(6,4))
plt.hist(mtcars["horsepower"], bins=12, color=(0, 0, 1, 0.3), edgecolor="blue")
plt.show()
```

##### Line Plot - Sepal Length vs Sepal Width


```python
# Line Plot (with Color Coding for Iris Dataset)
# Line Plot - Sepal Length vs Sepal Width
plt.figure(figsize=(7,5))
sns.lineplot(data=iris, x="sepal_length", y="sepal_width", hue="species")
plt.show()
```

##### Line Plot with Custom Colours


```python
# Custom colors
custom_palette = {"setosa": "darkorange", "versicolor": "green", "virginica": "red"}
plt.figure(figsize=(7,5))
sns.lineplot(data=iris, x="sepal_length", y="sepal_width", hue="species", palette=custom_palette)
plt.show()
```

**Plot 7:** Experiment with colours that are suitable for someone who is red-green colourblind.


```python
# Colourblind-friendly palette: avoid red and green together
custom_palette = {"setosa": "blue", "versicolor": "darkorange", "virginica": "purple"}
plt.figure(figsize=(7,5))
sns.lineplot(data=iris, x="sepal_length", y="sepal_width", hue="species", palette=custom_palette)
plt.show()
```
